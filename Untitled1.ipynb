```python
import pandas as pd
```

```python
sales = pd.Series([12000, 15000, 18000, 10000, 20000])
```

```python
#find highest, lowest, min,max,sum,avg
print(sales.sum())
print(sales.max())
print(sales.min())
print(sales.mean())
```

```python
marks = pd.Series([45, 78, 92, 56, 88, 67])
```

```python
print(marks.mean())
print(marks.max())
print(marks.min())
print(marks.count())
```

```python
print(marks[marks>70])
```

```python
print(marks[(marks >60) & (marks< 90)])
```

```python
marks = pd.Series([45, 78, 92, 56, 88, 67])
print(marks[(marks < 50) | (marks > 90)])
```

```python
print(marks[marks.isin([45,80,92])])
```
