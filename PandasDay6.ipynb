```python
import pandas as pd
```

```python
sales = pd.Series([15000,18000,22000,28000,300000])
sales
print(sales)
```

```python
print(sales.sum())
print(sales.mean())
print(sales.max())
print(sales.min())
print(sales.count())
```

```python
print(sales[sales>20000])
```

```python
print(sales[(sales >15000) & (sales <25000 )])
```

```python
sales = pd.Series([12000, 18000, 15000, 22000, 18000, 25000])
print(sales[sales.isin([12000, 22000, 25000])])
```

```python
salary = pd.Series([25000, 35000, None, 45000, None, 60000])
salary.isna()
```
