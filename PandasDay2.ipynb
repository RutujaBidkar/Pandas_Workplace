```python
import pandas as pd
```

```python
salary = pd.Series([25000,40000,50000,70000,900000])
print(salary[1])
```

```python
print(salary[(salary >30000) & (salary < 50000)])
```

```python
print(salary[((salary >30000) | (salary < 50000))])
```

```python
print(salary[~(salary > 40000)])
```

```python
Marks = pd.Series([45, 50, 90, 70, 85])

print(Marks.isin([45, 90]))
```

```python
print(Marks[Marks.isin([45,90])])
```

```python
print(Marks.where(Marks > 60))
```

```python
print(Marks.replace(90,100))
```

```python
print(Marks.sort_values(ascending = False))
```

```python
print(Marks.sort_values())
```

```python
s = pd.Series(
    [5000,9000,8000],
    index =["March","May","December"]
)
print(s.sort_index())
```

```python
print(s.sort_index(ascending = False))
```

```python
cities = pd.Series([
    "Pune", "Mumbai", "Pune",
    "Delhi", "Mumbai", "Pune"
])
print(cities.unique())
```
